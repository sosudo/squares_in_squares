In Squarecraft II, binary choices were sampled at ambiguous contacts. Here a preliminary fixed-angle LP supplies constraint shadow prices. ## 3. Dual-guided contact surgery
For each nearby pair, non-overlap requires at least one separating local axis. Freezing one axis makes centre compaction a linear program. Its shadow prices measure local objective sensitivity to those frozen constraints.

We distribute these pressures to incident squares and use them to choose a bounded set of binary axis alternatives. The subsequent MILP optimizes all centres and the box side jointly. We test two trust radii and allow a small initial projection deficit on alternative axes in the second pass. Pair-specific big-M constants come from the movement bounds; a radius of $\sqrt2+2\sqrt2t$ covers all pairs that might collide. Every returned candidate is independently repaired and checked before admission.

```python
"""Joint discrete contact-axis selection and continuous coordinate compaction."""
import numpy as np, math
from scipy.optimize import milp, linprog, Bounds, LinearConstraint
from scipy.spatial import cKDTree
from scipy.sparse import coo_matrix
from packing_geometry import normalize,side,repair

def guided_switch(p,trust=.22,seconds=3,seed=0,max_binary=160,axis_slack=0.):
    p=normalize(p);n=len(p);L=side(p);rng=np.random.default_rng(seed)
    a=p[:,2];c=np.cos(a);s=np.sin(a);h=(abs(c)+abs(s))/2
    ij=cKDTree(p[:,:2]).query_pairs(math.sqrt(2)+2*math.sqrt(2)*trust+1e-6,output_type='ndarray');i,j=ij.T;d=p[j,:2]-p[i,:2]
    ax=np.stack((np.column_stack((c[i],s[i])),np.column_stack((-s[i],c[i])),np.column_stack((c[j],s[j])),np.column_stack((-s[j],c[j]))),axis=1)
    pr=np.einsum('pk,pak->pa',d,ax);ax*=np.where(pr>=0,1,-1)[:,:,None];dist=abs(pr)
    rad=(1+abs(np.cos(a[i]-a[j]))+abs(np.sin(a[i]-a[j])))/2
    choices=[];amb=[]
    for k in range(len(i)):
        order=np.argsort(-dist[k]+rng.uniform(-1e-10,1e-10,4));opts=[int(order[0])]
        for t in order[1:]:
            if dist[k,t]<rad[k]-axis_slack-1e-7:continue
            if all(np.linalg.norm(ax[k,t]-ax[k,z])>1e-5 for z in opts):opts.append(int(t))
        choices.append(opts[:2])
        if len(opts)>1:amb.append(k)
    # LP shadow prices score a local neighbourhood of stressed contacts.
    primary=np.argmax(dist+rng.uniform(-1e-11,1e-11,dist.shape),axis=1)
    kk=np.arange(len(i));v=ax[kk,primary]
    cr=np.repeat(np.arange(len(i)),4);cidx=np.column_stack((2*i,2*i+1,2*j,2*j+1)).ravel()
    values=np.column_stack((v[:,0],v[:,1],-v[:,0],-v[:,1])).ravel()
    lr=[cr];lc=[cidx];lv=[values];rhs=[dist[kk,primary]-rad-2e-10];offset=len(i);k=np.arange(n)
    for dim in [0,1]:
        for upper in [False,True]:
            lr.append(np.repeat(np.arange(offset,offset+n),2));offset+=n
            lc.append(np.column_stack((2*k+dim,np.full(n,2*n))).ravel())
            lv.append(np.tile([1. if upper else -1.,-1. if upper else 0.],(n,1)).ravel())
            rhs.append(L-p[:,dim]-h if upper else p[:,dim]-h)
    mat=coo_matrix((np.concatenate(lv),(np.concatenate(lr),np.concatenate(lc))),shape=(offset,2*n+1)).tocsr()
    obj=np.zeros(2*n+1);obj[-1]=1
    lp=linprog(obj,A_ub=mat,b_ub=np.concatenate(rhs),bounds=[(-trust,trust)]*(2*n)+[(-trust,.02)],method='highs',options={'time_limit':1.,'primal_feasibility_tolerance':1e-9})
    weights=np.zeros(len(i));lpq=None
    if lp.success:
        weights=np.abs(lp.ineqlin.marginals[:len(i)])
        lpq=p.copy();lpq[:,:2]+=lp.x[:-1].reshape(n,2);lpq=repair(lpq,clearance=2e-9)
    node=np.zeros(n);np.add.at(node,i,weights);np.add.at(node,j,weights)
    contact_score=weights+.35*(node[i]+node[j])
    # A seeded component breaks degenerate ties without claiming stable dual prices.
    contact_score+=rng.random(len(i))*1e-9
    if len(amb)>max_binary:
        ranked=sorted(amb,key=lambda k:contact_score[k],reverse=True)
        keep=set(ranked[:max_binary//2]);other=[k for k in amb if k not in keep]
        keep.update(rng.choice(other,max_binary-len(keep),replace=False))
        choices=[v if k in keep else v[:1] for k,v in enumerate(choices)]
    off=2*n+1;num=0;rr=[];cc=[];vv=[];bb=[]
    def row(cols,vals,b):
        r=len(bb);rr.extend([r]*len(cols));cc.extend(cols);vv.extend(vals);bb.append(b)
    for k,(ii,jj) in enumerate(ij):
        opts=choices[k]
        for z,t in enumerate(opts):
            vx,vy=ax[k,t];cols=[2*ii,2*ii+1,2*jj,2*jj+1];vals=[vx,vy,-vx,-vy];rhs=dist[k,t]-rad[k]-1e-9
            if len(opts)>1:
                M=max(0,2*trust*(abs(vx)+abs(vy))-rhs)+1e-8
                cols.append(off+num);vals.append(M if z==0 else -M)
                if z==0:rhs+=M
            row(cols,vals,rhs)
        if len(opts)>1:num+=1
    for k in range(n):
        for dim in [0,1]:
            row([2*k+dim,2*n],[1,-1],L-p[k,dim]-h[k])
            row([2*k+dim],[-1],p[k,dim]-h[k])
    nv=off+num;A=coo_matrix((vv,(rr,cc)),shape=(len(bb),nv)).tocsc();objective=np.zeros(nv);objective[2*n]=1
    lo=np.r_[np.full(2*n,-trust),-trust,np.zeros(num)];hi=np.r_[np.full(2*n,trust),.02,np.ones(num)]
    result=milp(objective,integrality=np.r_[np.zeros(off),np.ones(num)],bounds=Bounds(lo,hi),constraints=LinearConstraint(A,-np.inf,bb),options={'time_limit':seconds,'mip_rel_gap':1e-6})
    if result.x is None:return lpq,{'status':int(result.status),'binaries':num,'lp_status':int(lp.status)}
    q=p.copy();q[:,:2]+=result.x[:2*n].reshape(n,2);q=repair(q,clearance=2e-9)
    if lpq is not None and (q is None or side(lpq)<side(q)):q=lpq
    return q,{'status':int(result.status),'binaries':num,'lp_status':int(lp.status),'stressed_contacts':int(np.sum(weights>1e-8))}
```

```python
"""Transfer new witnesses by trimming and adding integer L-shaped borders."""
import math,numpy as np
from packing_geometry import side,normalize,repair,aabbs

def propagate(bank, history, seeds, max_n=1000):
    for source in seeds:
        p=normalize(bank[source]); raw=p.copy();L=side(p);nn=len(p)
        # Deleting squares preserves validity; choose boundary removal by bbox.
        for _ in range(min(20,nn-5)):
            lo,hi=aabbs(raw);ext=set()
            for dim in [0,1]:
                ext.update(np.argsort(lo[:,dim])[:2]);ext.update(np.argsort(-hi[:,dim])[:2])
            choices=[np.delete(raw,k,axis=0) for k in ext]
            raw=min(choices,key=side);n=len(raw);ll=side(raw)
            if ll<side(bank[n])-1e-7:
                before=side(bank[n]);bank[n]=normalize(raw)
                history.append(dict(n=n,before=before,after=ll,method='subset transfer',parent=source))
        for ring in range(1,33):
            k=int(math.floor(L+1e-8));extra=np.array([(j+.5,L+.5,0) for j in range(k)]+[(L+.5,j+.5,0) for j in range(k)]+[(L+.5,L+.5,0)])
            p=np.vstack((p,extra));L+=1
            if len(p)>max_n+20:break
            if len(p)>max_n:continue
            # All last 20 counts can inherit a subset of this enlarged witness.
            for n in range(max(source,len(p)-20),len(p)+1):
                q=p[:n];ll=side(q)
                if ll<side(bank[n])-1e-7:
                    q=repair(q,clearance=2e-9);ll=side(q)
                    if ll<side(bank[n])-1e-7:
                        before=side(bank[n]);bank[n]=q.copy();history.append(dict(n=n,before=before,after=ll,method='border transfer',parent=source))
    return bank
```

## 5. CPU search with stage-by-stage attribution
The baseline is freshly loaded before search. A strictly smaller enclosure is required for admission. Promising MILP candidates receive free-angle sparse LP refinement inherited from the credited CPU implementation. Improvements are transferred by subset selection and integer borders.

Fixed seeds determine sampling. Time-limited solver termination may differ across machines. Solver status, binary counts, LP status and the number of stressed contacts are recorded for every attempt. A failed attempt does not prove that no improvement exists.

```python
started=time.time()
active=[n for n in range(5,1001) if side(bank[n])<math.ceil(math.sqrt(n))-1e-7]
for epoch,(trust,slack) in enumerate([(.3,0.),(.55,.12)]):
    for ix,n in enumerate(active):
        before=side(bank[n])
        q,info=guided_switch(bank[n],seed=81024+epoch*10000+n,trust=trust,axis_slack=slack,seconds=.65,max_binary=160)
        attempts.append(dict(n=n,epoch=epoch,method='dual-guided contact surgery',**info))
        if q is not None and side(q)<before-1e-7:
            q,trace,_=polish(q,rounds=35,max_seconds=3)
            after=side(q)
            if after<before-1e-7:
                bank[n]=q;record=dict(n=n,before=before,after=after,gain=(before-after)/math.sqrt(n),method='dual-guided contact surgery',epoch=epoch)
                history.append(record);print('IMPROVED',json.dumps(record),flush=True)
        if ix%40==0:print('SURGERY',epoch,ix,len(active),total_score(bank),round(time.time()-started,1),flush=True)
    stages.append(dict(stage=f'Dual-guided contact surgery pass {epoch+1}',score=total_score(bank)))
    np.savez_compressed('surgery_bank.npz',**{str(n):p for n,p in bank.items()})
    Path('search_history.json').write_text(json.dumps(history,indent=2))
    Path('search_attempts.json').write_text(json.dumps(attempts,indent=2))
seeds=sorted({h['n'] for h in history})
propagate(bank,history,seeds)
stages.append(dict(stage='Subset and border transfer',score=total_score(bank)))
np.savez_compressed('search_bank.npz',**{str(n):p for n,p in bank.items()})
Path('search_history.json').write_text(json.dumps(history,indent=2))
print('SEARCH COMPLETE',total_score(bank),'accepted steps',len(history),'seconds',time.time()-started)
```

```python
import mpmath
if not hasattr(mpmath.mpf(1),'man'):
    subprocess.run([sys.executable,'-m','pip','install','-q','mpmath==1.4.1'],check=True)
sys.path.insert(0,str(ROOT/'reference_sources'))
import official_metric as metric
from collections import defaultdict
base_rows=defaultdict(list)
with open('baseline.csv',newline='') as f:
    for row in csv.DictReader(f):base_rows[int(row['n'])].append(row)
rows_out={};certified=[];rejected_changes=[]
for n in range(1,1001):
    if side(bank[n])>=side(initial[n])-1e-7:
        if n in base_rows:rows_out[n]=base_rows[n]
        continue
    q=normalize(bank[n]);q[:,:2]*=1.00000001
    rows=[dict(n=str(n),square_id=str(k),x=format(x,'.17g'),y=format(y,'.17g'),angle=format(a,'.17g')) for k,(x,y,a) in enumerate(q)]
    result=metric._evaluate_puzzle_worker((n,rows))
    baseline_n=side(initial[n])/math.sqrt(n)-1
    if result[3]=='ok' and result[1]<baseline_n-1e-9:
        rows_out[n]=rows;certified.append(n)
    else:
        rejected_changes.append(dict(n=n,result=result))
        if n in base_rows:rows_out[n]=base_rows[n]
        bank[n]=initial[n].copy()
with open('submission.csv','w',newline='') as f:
    w=csv.writer(f);w.writerow(['id','n','square_id','x','y','angle']);idx=0
    for n,rows in sorted(rows_out.items()):
        for k,row in enumerate(rows):
            w.writerow([idx,n,k,row['x'],row['y'],row['angle']]);idx+=1
subprocess.run([sys.executable,'validate_submission.py','submission.csv','validation.json','--workers','2'],check=True)
pd.read_csv('submission.csv').to_csv('float64_roundtrip.csv',index=False,float_format='%.17g')
subprocess.run([sys.executable,'validate_submission.py','float64_roundtrip.csv','roundtrip_validation.json','--workers','2'],check=True)
report=json.loads(Path('validation.json').read_text());roundtrip=json.loads(Path('roundtrip_validation.json').read_text())
assert abs(report['score']-roundtrip['score'])<1e-8
assert report['score']<=baseline_score+1e-8
stages.append(dict(stage='Official decimal certification',score=report['score']))
summary=dict(score=report['score'],baseline_score=baseline_score,incremental_gain=baseline_score-report['score'],certified_improved_sizes=certified,rejected_changes=rejected_changes,rows=idx,sha256=report['sha256'],seconds=time.time()-started)
Path('run_summary.json').write_text(json.dumps(summary,indent=2))
pd.DataFrame(stages).to_csv('ablation.csv',index=False)
display(pd.DataFrame(stages));print(json.dumps(summary,indent=2))
```
